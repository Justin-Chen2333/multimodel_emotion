```python
import librosa
import librosa.display
```

```python
r = df.iloc[0]
wav = DATA / "audio" / "train" / f"dia{r.Dialogue_ID}_utt{r.Utterance_ID}.wav"
y, sr = librosa.load(wav, sr=16000)

print(r.Utterance, "|", r.Emotion)
print("采样率:", sr, " 采样点数:", len(y), " 时长（秒）:", len(y) / sr)
```

```python
plt.figure(figsize=(10, 3))
librosa.display.waveshow(y, sr=sr)
plt.title(f"{r.Utterance} [{r.Emotion}]")
plt.show()
```

```python
from IPython.display import Audio
Audio(y, rate=sr)
```
